```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["YearMonth"]=df["Order Date"].dt.to_period("M").astype(str)
monthly_revenue=df.groupby(df["YearMonth"])["Price (INR)"].sum().reset_index()

plt.figure()
plt.plot(monthly_revenue["YearMonth"],monthly_revenue["Price (INR)"])
plt.xticks(rotation=45)
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")
plt.title("Monthly Sales Trend")
plt.tight_layout()
plt.show()
```

### Quartely Performance Summary

```python
df["Order Date"]=pd.to_datetime(df["Order Date"])
df["Quarter"]=df["Order Date"].dt.to_period("Q").astype(str)
quertly_summary=df.groupby(df["Quarter"]).agg(
Total_Sales=("Price (INR)","sum"),
Avg_Rating=("Rating","mean"),
Total_Orders=("Order Date","count")
).sort_values("Quarter")  

quertly_summary["Total_Sales"]=quertly_summary["Total_Sales"].round(0)
quertly_summary["Avg_Rating"]=quertly_summary["Avg_Rating"].round(2)
```

```python
plt.plot(quertly_summary.index,
         quertly_summary["Total_Sales"],
         marker="o"
    
)
plt.xlabel("Quarter")
plt.ylabel("total_Sales")
plt.title("Quarterly Sales Trend")
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

# Total Sales - Bar
bars = ax1.bar(
    quartely_summary.index,
    quartely_summary["Total_Sales"],
    label="Total Sales"
)

ax1.set_xlabel("Quarter")
ax1.set_ylabel("Total Sales (INR)")
ax1.bar_label(bars, fmt="%.0f")

# Total Orders - Line
ax2 = ax1.twinx()

ax2.plot(
    quartely_summary.index,
    quartely_summary["Total_orders"],
    marker="o",
    label="Total Orders"
)

ax2.set_ylabel("Total Orders")

plt.title("Quarterly Sales and Orders")
plt.show()
```
